## Perform Simple Regression

### Subtask:
Perform a simple linear regression with 'FemaleLabor' as the dependent variable and 'HDI' as the independent variable.


**Reasoning**:
To perform a simple linear regression, I will first define the dependent ('FemaleLabor') and independent ('HDI') variables. I will then combine them into a DataFrame to ensure consistent handling of missing values using `dropna()`. After that, I will add a constant to the independent variable, fit the OLS model, and print its summary, following the instructions provided.



````python
Y = df['FemaleLabor']
X = df['HDI']

# Combine Y and X into a single DataFrame to drop NaNs consistently
data_for_reg = pd.DataFrame({'FemaleLabor': Y, 'HDI': X}).dropna()

Y = data_for_reg['FemaleLabor']
X = data_for_reg['HDI']

X = sm.add_constant(X)

reg_equation = sm.OLS(Y, X).fit()
print(reg_equation.summary())
````
